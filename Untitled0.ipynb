```python
import warnings
warnings.filterwarnings("ignore")
import gc, json, re
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from collections import Counter
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, AutoModelForCausalLM
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
import torch
import torch.nn.functional as F
from tqdm import tqdm

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Urządzenie: {device}")

LABEL2ID = {"plus": 0, "minus": 1, "neutral": 2}
ID2LABEL = {0: "plus", 1: "minus", 2: "neutral"}
LABEL_NAMES = ["plus", "minus", "neutral"]


def map_to_sentiment(label_str):
    s = str(label_str).lower().strip()
    if any(x in s for x in ["plus", "positive", "pozytyw"]): return "plus"
    if any(x in s for x in ["minus", "negative", "negatyw"]): return "minus"
    if any(x in s for x in ["zero", "neutral"]): return "neutral"
    if any(x in s for x in ["amb"]): return "ambiguous"
    return None


def evaluate(y_true, y_pred, title="Model"):
    pairs = [(t, p) for t, p in zip(y_true, y_pred) if p is not None]
    if not pairs:
        print(f"  {title}: brak poprawnych predykcji!"); return None
    yt, yp = zip(*pairs)
    acc = accuracy_score(yt, yp)
    f1m = f1_score(yt, yp, average="macro", zero_division=0)
    f1w = f1_score(yt, yp, average="weighted", zero_division=0)
    print(f"\n--- {title} ---")
    print(f"Accuracy: {acc:.4f}  |  F1 macro: {f1m:.4f}  |  F1 weighted: {f1w:.4f}  |  Pokrycie: {len(yt)}/{len(y_pred)}")
    print(classification_report(yt, yp, target_names=LABEL_NAMES, zero_division=0))
    return {"title": title, "accuracy": acc, "f1_macro": f1m, "f1_weighted": f1w, "coverage": len(yt)/len(y_pred)}

```

```python
dataset = load_dataset("allegro/klej-polemo2-in")
for split in dataset:
    print(f"  {split}: {len(dataset[split])} próbek")

test_data = dataset["test"]

print(f"\nUnikalne etykiety w zbiorze testowym: {set(test_data['target'])}")


mapped = [(s, map_to_sentiment(t)) for s, t in zip(test_data["sentence"], test_data["target"])]
mapped = [(s, l) for s, l in mapped if l != "ambiguous"]
texts = [s for s, _ in mapped]
labels_clean = [l for _, l in mapped]
true_labels = [LABEL2ID[l] for l in labels_clean]

print(f"Po usunięciu ambiguous: {len(texts)} próbek")


counts = Counter(labels_clean)
print("Rozkład klas:")
for l in LABEL_NAMES:
    c = counts.get(l, 0)
    print(f"  {l}: {c} ({c/len(texts)*100:.1f}%)")


lens = [len(t.split()) for t in texts]
print(f"\nDługość tekstów (słowa): średnia={np.mean(lens):.0f}, mediana={np.median(lens):.0f}, "
      f"min={min(lens)}, max={max(lens)}")


for l in LABEL_NAMES:
    ll = [lens[i] for i, c in enumerate(labels_clean) if c == l]
    print(f"  {l}: średnia={np.mean(ll):.0f}, mediana={np.median(ll):.0f}")
```

```python
ENCODER_MODEL_1 = "Voicelab/herbert-base-cased-sentiment"
print(f"Model: {ENCODER_MODEL_1}")

tokenizer1 = AutoTokenizer.from_pretrained(ENCODER_MODEL_1)
model1 = AutoModelForSequenceClassification.from_pretrained(ENCODER_MODEL_1).to(device)
model1.eval()
print(f"Etykiety modelu: {model1.config.id2label}")


id2label = model1.config.id2label
valid_idx = []   # indeksy klas modelu które nas interesują
valid_tgt = []   # odpowiadające im indeksy klas zbioru
for mid, mlabel in id2label.items():
    mapped = map_to_sentiment(mlabel)
    if mapped and mapped in LABEL2ID:
        valid_idx.append(int(mid))
        valid_tgt.append(LABEL2ID[mapped])
print(f"Mapowanie indeksów modelu {valid_idx} -> indeksów zbioru {valid_tgt}")


def predict_encoder(model, tokenizer, texts, temperature=1.0, batch_size=32):
    model.eval()
    preds = []
    for i in range(0, len(texts), batch_size):
        batch = texts[i:i+batch_size]
        enc = tokenizer(batch, padding=True, truncation=True, max_length=512, return_tensors="pt").to(device)
        with torch.no_grad():
            logits = model(**enc).logits[:, valid_idx] / temperature
        local_pred = torch.argmax(logits, dim=-1)
        preds.extend([valid_tgt[p] for p in local_pred.cpu().tolist()])
    return preds


enc1_preds = predict_encoder(model1, tokenizer1, texts)
results_enc1 = evaluate(true_labels, enc1_preds, f"Encoder: herbert-sentiment")

```

```python
print("\n" + "=" * 60)
print("PUNKT 3: Eksploracja parametrów encoder")
print("=" * 60)

# --- Aspekt 1: temperatura softmax ---
print("\n>>> Aspekt 1: Wpływ temperatury softmax")
temperatures = [0.25, 0.5, 1.0, 2.0, 3.0]
temp_res = {}
for t in temperatures:
    p = predict_encoder(model1, tokenizer1, texts, temperature=t)
    r = evaluate(true_labels, p, f"Encoder herbert T={t}")
    temp_res[t] = r

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(temperatures, [temp_res[t]["accuracy"] for t in temperatures], "o-", label="Accuracy")
ax.plot(temperatures, [temp_res[t]["f1_macro"] for t in temperatures], "s-", label="F1 macro")
ax.set_xlabel("Temperatura"); ax.set_ylabel("Wynik"); ax.set_title("Temperatura softmax – encoder")
ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout(); plt.savefig("enc_temperature.png", dpi=120); plt.show()

del model1, tokenizer1; gc.collect()
if torch.cuda.is_available(): torch.cuda.empty_cache()

# --- Aspekt 2: inny model encoder ---
print("\n>>> Aspekt 2: Drugi model encoder")
ENCODER_MODEL_2 = "cardiffnlp/twitter-xlm-roberta-base-sentiment"
print(f"Model 2: {ENCODER_MODEL_2}")

tokenizer2 = AutoTokenizer.from_pretrained(ENCODER_MODEL_2)
model2 = AutoModelForSequenceClassification.from_pretrained(ENCODER_MODEL_2).to(device)
model2.eval()
print(f"Etykiety modelu 2: {model2.config.id2label}")

# Nowe mapowanie dla modelu 2
id2label2 = model2.config.id2label
valid_idx2, valid_tgt2 = [], []
for mid, mlabel in id2label2.items():
    mapped = map_to_sentiment(mlabel)
    if mapped and mapped in LABEL2ID:
        valid_idx2.append(int(mid)); valid_tgt2.append(LABEL2ID[mapped])

# Nadpisanie globali tymczasowo
_vi, _vt = valid_idx, valid_tgt
valid_idx, valid_tgt = valid_idx2, valid_tgt2

enc2_preds = predict_encoder(model2, tokenizer2, texts)
results_enc2 = evaluate(true_labels, enc2_preds, f"Encoder: xlm-roberta-sentiment")

valid_idx, valid_tgt = _vi, _vt  # przywrócenie

del model2, tokenizer2; gc.collect()
if torch.cuda.is_available(): torch.cuda.empty_cache()

# Porównanie encoderów
print("\nPorównanie encoderów:")
for r in [results_enc1, results_enc2]:
    if r: print(f"  {r['title']:<40} Acc={r['accuracy']:.4f}  F1={r['f1_macro']:.4f}")
```

```python
print("\n" + "=" * 60)
print("PUNKT 4: Klasyfikacja decoder-only")
print("=" * 60)

DECODER_MODEL = "Qwen/Qwen2.5-1.5B-Instruct"
print(f"Model: {DECODER_MODEL}")

tokenizer_dec = AutoTokenizer.from_pretrained(DECODER_MODEL)
model_dec = AutoModelForCausalLM.from_pretrained(
    DECODER_MODEL,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32,
    device_map="auto" if device == "cuda" else None,
)
if device != "cuda":
    model_dec.to(device)
model_dec.eval()

PROMPT_SIMPLE = (
    "Jesteś ekspertem od analizy wydźwięku polskich recenzji.\n"
    "Sklasyfikuj recenzję jako: plus (pozytywna), minus (negatywna), neutral (neutralna).\n"
    "Odpowiedz WYŁĄCZNIE jednym słowem: plus, minus lub neutral."
)


def parse_response(text):
    """Parsuje wyjście decodera na etykietę."""
    s = text.strip().lower()
    # JSON
    try:
        m = re.search(r"\{[^}]+\}", s)
        if m:
            d = json.loads(m.group())
            for v in d.values():
                r = map_to_sentiment(str(v))
                if r and r in LABEL2ID: return r
    except: pass
    # Pierwsze słowo
    w = re.sub(r"[^a-ząćęłńóśżź]", "", s.split()[0]) if s.split() else s
    r = map_to_sentiment(w)
    if r and r in LABEL2ID: return r
    # Cały tekst
    r = map_to_sentiment(s)
    if r and r in LABEL2ID: return r
    return None


def classify_decoder(model, tokenizer, texts, system_prompt, temperature=0.1, max_new_tokens=10):
    """Klasyfikacja zero-shot modelem decoder."""
    model.eval()
    preds, raw = [], []
    dev = next(model.parameters()).device
    for i in tqdm(range(len(texts)), desc="Decoder"):
        messages = [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": f"Sklasyfikuj wydźwięk recenzji:\n\n{texts[i]}"},
        ]
        prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=2048).to(dev)
        with torch.no_grad():
            gen_kw = dict(max_new_tokens=max_new_tokens, pad_token_id=tokenizer.eos_token_id)
            if temperature > 0:
                gen_kw.update(temperature=temperature, do_sample=True, top_p=0.9)
            else:
                gen_kw.update(do_sample=False)
            out = model.generate(**inputs, **gen_kw)
        resp = tokenizer.decode(out[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True)
        raw.append(resp)
        label = parse_response(resp)
        preds.append(LABEL2ID[label] if label else None)
    return preds, raw


print(f"\nKlasyfikacja {len(texts)} próbek (T=0.1)…")
dec_preds, dec_raw = classify_decoder(model_dec, tokenizer_dec, texts, PROMPT_SIMPLE, temperature=0.1)

# Przykłady
print("\nPrzykładowe wyjścia:")
for i in range(min(5, len(dec_raw))):
    print(f"  [{i}] '{dec_raw[i]}' → {ID2LABEL.get(dec_preds[i], 'BRAK')} (prawda: {ID2LABEL[true_labels[i]]})")

n_fail = sum(1 for p in dec_preds if p is None)
if n_fail: print(f"Niesparsowane: {n_fail}/{len(dec_preds)}")

results_dec_base = evaluate(true_labels, dec_preds, f"Decoder: Qwen2.5-1.5B (prosty, T=0.1)")
```

```python
print("\n" + "=" * 60)
print("PUNKT 5: Eksploracja parametrów decoder")
print("=" * 60)

# --- Aspekt 1: Różne prompty ---
print("\n>>> Aspekt 1: Porównanie promptów")

PROMPT_JSON = (
    "Jesteś ekspertem od analizy wydźwięku polskich recenzji.\n"
    "Sklasyfikuj recenzję jako: plus, minus, neutral.\n"
    'Odpowiedz WYŁĄCZNIE w formacie JSON: {"label": "<etykieta>"}\n'
    "Nie dodawaj wyjaśnień."
)

PROMPT_FEWSHOT = (
    "Klasyfikuj wydźwięk recenzji jako: plus, minus, neutral.\n\n"
    "Przykłady:\n"
    '"Świetny hotel, polecam!" → plus\n'
    '"Okropna obsługa, nie polecam." → minus\n'
    '"Hotel w centrum, pokoje standardowe." → neutral\n\n'
    "Odpowiedz jednym słowem: plus, minus lub neutral."
)

prompt_results = {"prosty": results_dec_base}
for name, (prompt, max_tok) in {"JSON": (PROMPT_JSON, 20), "few-shot": (PROMPT_FEWSHOT, 10)}.items():
    print(f"\n  Prompt: {name}")
    p, _ = classify_decoder(model_dec, tokenizer_dec, texts, prompt, temperature=0.1, max_new_tokens=max_tok)
    prompt_results[name] = evaluate(true_labels, p, f"Decoder ({name}, T=0.1)")

print("\nPorównanie promptów:")
for n, r in prompt_results.items():
    if r: print(f"  {n:<12} Acc={r['accuracy']:.4f}  F1={r['f1_macro']:.4f}  Pokr={r['coverage']*100:.0f}%")

# --- Aspekt 2: Różne temperatury ---
print("\n>>> Aspekt 2: Wpływ temperatury generacji")
dec_temps = [0.01, 0.1, 0.5, 0.7, 1.0]
dec_temp_res = {}
for t in dec_temps:
    print(f"\n  T={t}")
    p, _ = classify_decoder(model_dec, tokenizer_dec, texts, PROMPT_SIMPLE, temperature=t)
    dec_temp_res[t] = evaluate(true_labels, p, f"Decoder T={t}")

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(dec_temps, [dec_temp_res[t]["accuracy"] if dec_temp_res[t] else 0 for t in dec_temps], "o-", label="Accuracy")
ax.plot(dec_temps, [dec_temp_res[t]["f1_macro"] if dec_temp_res[t] else 0 for t in dec_temps], "s-", label="F1 macro")
ax.set_xlabel("Temperatura"); ax.set_ylabel("Wynik"); ax.set_title("Temperatura generacji – decoder")
ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout(); plt.savefig("dec_temperature.png", dpi=120); plt.show()

# --- Aspekt 3: Kwantyzacja ---
print("\n>>> Aspekt 3: Kwantyzacja")
del model_dec; gc.collect()
if torch.cuda.is_available(): torch.cuda.empty_cache()

results_quant = None
try:
    from transformers import BitsAndBytesConfig
    print(f"Ładowanie {DECODER_MODEL} z kwantyzacją 4-bit…")
    model_q = AutoModelForCausalLM.from_pretrained(
        DECODER_MODEL,
        quantization_config=BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_compute_dtype=torch.float16, bnb_4bit_quant_type="nf4"),
        device_map="auto",
    )
    model_q.eval()
    pq, _ = classify_decoder(model_q, tokenizer_dec, texts, PROMPT_SIMPLE, temperature=0.1)
    results_quant = evaluate(true_labels, pq, "Decoder Qwen 4-bit (T=0.1)")
    del model_q; gc.collect()
    if torch.cuda.is_available(): torch.cuda.empty_cache()
except Exception as e:
    print(f"Kwantyzacja 4-bit niedostępna: {e}\nPomijanie.")
```

```python
print("\n" + "=" * 60)
print("PODSUMOWANIE KOŃCOWE")
print("=" * 60)

all_results = [r for r in [
    results_enc1, results_enc2, results_dec_base,
    *prompt_results.values(), results_quant,
    *dec_temp_res.values(),
] if r is not None]

# Deduplikacja po tytule
seen = set()
unique_results = []
for r in all_results:
    if r["title"] not in seen:
        seen.add(r["title"])
        unique_results.append(r)

print(f"\n{'Model':<50} {'Acc':>7} {'F1m':>7} {'F1w':>7}")
print("-" * 73)
for r in unique_results:
    print(f"{r['title']:<50} {r['accuracy']:>7.4f} {r['f1_macro']:>7.4f} {r['f1_weighted']:>7.4f}")

# Wykres porównawczy
if unique_results:
    fig, ax = plt.subplots(figsize=(12, 6))
    names = [r["title"] for r in unique_results]
    x = np.arange(len(names))
    w = 0.35
    ax.bar(x - w/2, [r["accuracy"] for r in unique_results], w, label="Accuracy", color="steelblue")
    ax.bar(x + w/2, [r["f1_macro"] for r in unique_results], w, label="F1 macro", color="coral")
    ax.set_xticks(x); ax.set_xticklabels(names, rotation=45, ha="right", fontsize=7)
    ax.set_ylabel("Wynik"); ax.set_title("Porównanie: encoder vs decoder"); ax.legend()
    ax.set_ylim(0, 1.1); ax.grid(axis="y", alpha=0.3)
    plt.tight_layout(); plt.savefig("final_comparison.png", dpi=120, bbox_inches="tight"); plt.show()

print("\n✓ Gotowe!")
```
